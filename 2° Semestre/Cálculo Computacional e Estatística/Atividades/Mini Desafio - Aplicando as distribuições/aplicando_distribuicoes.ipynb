# Criando uma quantidade padrão de análise

A partir de uma discussão, será utilizado uma amostragem de 200 pessoas para realizar todas as análises e observações.

# Como foi feito

Para criar o número padrão representado a amostragem, foi criado uma varíavel quantitativa discreta, com o valor de 200.

```r
numero_amostragem <- 200
```

# Simulando as idades dos Jogadores

A partir de um palpite dos desenvolvedores do jogo, a idade média entre os jogadores é 25 anos, com uma variação de 6 anos, por causa que os jogadores podem ser pessoas mais novas e mais velhas. Além disso, como a análise foi feita a partir de uma média, a utilização da distruição normal é a que mais se encaixa para este cenário.

# Como foi feito

Utilizando a função rnorm(), utilizarei a variável que representa a amostra da parcela do meu total de pessoas, além de utilizar as informações e palpites feitos pelos desenvolvedores.

Para melhor visualização das idades, foi criado um histograma, já que o histograma é a melhor opção para a análise de variáveis quantitativas.

```r
idade.jogadores <- trunc(rnorm(n = numero_amostragem, mean = 25, sd = 6))
```

```r
hist(idade.jogadores,
     main = "Quantidade idades dos Jogadores",
     xlab = "Quantidade",
     ylab = "Idades"
     )


axis(side = 1, at = seq(10, 40, by = 5))
```

# Simulando qunatos Jogadores Visitam a Loja por Hora

Como durante uma hora, a quantidade de jogadores que entra na lógica é muito variado, entretando, em horário de pico a média de jogadores que entre na loja é de 30 jogadores, por hora. Além disso, para realizar esta simulação, é necessário utilizar a distruição poisson, já que é necessário saber a quantidade de ocorrência em um determinado intervalo de tempo.

# Como foi feito

A partir da análise e necessidade de "randomizar" alguns valores para amostrar a partir de uma distruição de poisson, foi utilizado a função rpois() para sanar esta dificuldade. Utilizando os parâmetros: 8 (para representar a quantidade de valores/amostrar para serem geradas) e 30 (para representar a taxa média ou média esperada de ocorrência dos eventos por intervalo). Além de utilizar a função trunc() para gerar valores inteiros, ou seja, sem casa decimal.

```r
quantidade.jogadores.hora <- trunc(rpois(n = 8, lambda = 30))

barplot(quantidade.jogadores.hora,
        main = "Quantidade de jogadores por hora",
        names.arg = c(1:8),
        xlab = "Horários",
        ylab = "Quantidade de Jogadores")
```

# Simulando o Uso de Cupons Virtuais

A loja de jogos distruiu cerca de 500 cupons para os seus jogadores. Em uma análise, foi calculado que um jogador possui 45% de chance de utilizar um cupom. Desta forma, para realizar uma análise de um fenômeno ser verdadeiro, foi utilizado a distruição binominal.

# Como foi feito

Utilizando a função rbinom(), para fazer uma "randominzação" algumas amostrar a partir da distruição binominal, foi utilizado os paramêtros: numero_amostragem (para saber a quantidade de registros gerados), 1 (para representar se vai ser usado ou não o cupom) e a probabilidade de utilizar este cupom, sendo 45%.

```r
probabilidade.usar.cupom <- rbinom(
      numero_amostragem,
       1,
       0.45)

barplot(table(probabilidade.usar.cupom),
              names.arg = c("Não", "Sim"),
              col = c("red", "green"),
              main = "Jogadores Usaram Cupom?",
              )
```
